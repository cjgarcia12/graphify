# Daily revenue


```python
catalog = dbutils.widgets.get("catalog")

def compute():
    orders = spark.table(f"{catalog}.silver.orders_enriched")
    return orders.groupBy("order_date").sum("amount")
```

```python
compute().write.mode("overwrite").saveAsTable(f"{catalog}.gold.daily_revenue")
```

```python
%sql
SELECT order_date, `sum(amount)` FROM gold.daily_revenue ORDER BY 1 DESC LIMIT 10
```
